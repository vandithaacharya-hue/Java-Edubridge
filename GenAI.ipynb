#Step 3: Improting Libraries

```python
from transformers import pipeline
```

#Step 4: Load SLM (Small Language Model)

```python
generator = pipeline(
    'text-generation',
    model= "distilgpt2"
)
```

#Step 5: Generate First AI Text

```python
result = generator(
    "Artificial intelligence is",
    max_new_tokens=80

)
print(result[0]["generated_text"])
```

#Step 6: Change the Prompt

```python
result = generator(
    "Generative Artificial Intelligence is a domian of AI",
    max_new_tokens=50

)
print(result[0]["generated_text"])
```

#Step 7: Create our Own Function

```python
def generate_text(prompt):

  result = generator(
      prompt,
      max_new_tokens = 50
  )

  return result[0]["generated_text"]

print(generate_text("A CIT Student uses AI for"))
```

#Step 8: Build Mini AI Application

```python
topic = input("Enter a topic:")
prompt = f"{topic} is"
print(generate_text(prompt))
```
